# Graus de liberdade

Grau de liberdade é, em estatística, o número de determinações independentes (dimensão da amostra) menos o número de parâmetros estatísticos a serem avaliados na população. Por exemplo, para estimar a variância de uma amostra de *N* observações, e sua estimativa depende da média da amostra, então o grau de liberdade é *N - 1*.

## Exemplo 1

Em um programa de jogos onde um prêmio está escondido atrás de uma das três portas, você precisaria abrir duas portas para ter certeza de onde o prêmio está localizado.

1. **Situação inicial:** Você tem três portas (Porta A, Porta B e Porta C) e o prêmio está escondido atrás de uma delas.
2. **Primeira abertura:** Se você abrir uma porta e não encontrar o prêmio, há duas portas restantes.
3. **Segunda abertura:** Se você abrir uma segunda porta e ainda não encontrar o prêmio, a única porta restante deve conter o prêmio.

Portanto, ao abrir duas das três portas, você poderá determinar com certeza onde o prêmio está localizado.

## Exemplo 3

Suponha que você tenha quatro variáveis de um retângulo:

- o comprimento,
- a largura,
- a área
- e o perímetro de um retângulo.

Você realmente conhece quatro coisas? 

Não, porque há apenas dois graus de liberdade. 

- Se você conhece o comprimento e a largura, pode derivar a área e o perímetro.
- Se você conhece o comprimento e a área, pode derivar a largura e o perímetro.
- Se você conhece a área e o perímetro, pode derivar o comprimento e a largura.
- Se você tem todas as quatro variáveis, pode dizer que o sistema é consistente (todas as variáveis concordam entre si) ou inconsistente (nenhum retângulo poderia realmente satisfazer todas as condições).

Um quadrado é um retângulo com um grau de liberdade removido; se você conhece qualquer lado de um quadrado ou seu perímetro ou sua área, pode derivar todos os outros porque há apenas um grau de liberdade.

Em estatística a ideia é a mesma. Se todos os dados que você está usando como entrada para uma função são variáveis independentes, então você tem tantos graus de liberdade quanto tem entradas. Mas se elas têm alguma dependência de alguma forma, de modo que se você tivesse $n - k$ entradas poderia descobrir os $k$ restantes, então na verdade você só tem $n - k$ graus de liberdade. 

## Como saber o valor?

Um conjunto de dados contém um número de observações, digamos, $ n $. Elas constituem $ n $ registros. Esses registros podem ser usadas para estimar parâmetros ou variabilidade. Em geral, cada item a ser estimado custa um grau de liberdade. Os graus de liberdade restantes são usados para estimar a variabilidade. Tudo o que precisamos fazer é contar corretamente.

**Uma única amostra:** Há $ n $ observações. Há um parâmetro (a média) que precisa ser estimado. Isso deixa $ n - 1 $ graus de liberdade para estimar a variabilidade.

**Duas amostras:** Há $ n_1 + n_2 $ observações. Há duas médias a serem estimadas. Isso deixa $ n_1 + n_2 - 2 $ graus de liberdade para estimar a variabilidade.

**ANOVA com $ g $ grupos:** Há $ n_1 + \ldots + n_g $ observações. Há $ g $ médias a serem estimadas. Isso deixa $ n_1 + \ldots + n_g - g $ graus de liberdade para estimar a variabilidade. Isso explica os graus de liberdade para a estatística F.

A hipótese nula primária testada pela ANOVA é que as $ g $ médias populacionais são iguais. A hipótese nula é que há uma única média. A hipótese alternativa é que há $ g $ médias individuais. Portanto, há $ g - 1 $ -- isto é, $ g $ ( $ H_1 $ ) menos 1 ( $ H_0 $ ) -- graus de liberdade para testar a hipótese nula.

**Regressão múltipla com $ p $ preditores:** Há $ n $ observações com $ p + 1 $ parâmetros a serem estimados -- um coeficiente de regressão para cada um dos preditores mais o intercepto. Isso deixa $ n - p - 1 $ graus de liberdade para erro.
