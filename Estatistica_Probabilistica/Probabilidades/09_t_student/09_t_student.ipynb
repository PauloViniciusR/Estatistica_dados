# Distribuições de probabilidade para variáveis contínuas

## Distribuição *t* de *Student*

A distribuição *t* de *Student* é uma das principais distribuições de probabilidade, com inúmeras aplicações em inferência estatística.

Vamos supor uma variável aleatória *Z* que tenha distribuição normal com média zero e desvio-padrão 1, e uma variável aleatória *X* com distribuição qui-quadrado com $\nu$ graus de liberdade, de modo que *Z* e *X* sejam independentes. Uma variável aleatória contínua *T* pode então ser definida como:

$$
T = \frac{Z}{\sqrt{\frac{X}{\nu}}}
$$

Podemos dizer que a variável *T* possui uma distribuição *t* de *Student* com $\nu$ graus de liberdade se sua função densidade de probabilidade for dada por:

$$
f(t) = \frac{ \Gamma \left( \frac{\nu + 1}{2} \right) }{ \Gamma \left( \frac{\nu}{2} \right) \cdot \sqrt{\pi \nu} } \cdot \left( 1 + \frac{t^2}{\nu} \right)^{-\frac{\nu + 1}{2}}, -\infty < t< \infty
$$

em que $\Gamma (\alpha) = \int\limits_{0}^{\infty} e^{-x} \cdot x^{\alpha - 1} dx $

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm, t

for graus_liberdade in range(1, 10):
    x = np.linspace(-4, 4, 10_000)
    y = t.pdf(x, graus_liberdade)

    plt.plot(x, y, label=f"Graus de liberdade : {graus_liberdade}")

x_norm = np.linspace(-4, 4, 10_000)
y_norm = norm.pdf(x_norm, 0, 1)

plt.plot(x_norm, y_norm, label="Normal", color="black")

plt.xlim(-4, 4)
plt.legend()

```

Podemos notar que a distribuição t de Student é simétrica em torno da média, com formato de sino, e assemelha-se a uma distribuição normal padrão, porém, com caudas mais largas, podendo gerar valores mais extremos que aqueles presentes em uma distribuição normal.

O parâmetro $\nu$ (número de graus de liberdade) define e caracteriza a forma da distribuição t de Student; quanto maior for o valor de $\nu$, mais a distribuição t de Student se aproxima de uma normal padrão.

Assim como as distribuições normal e qui-quadrado, a distribuição t de Student tem inúmeras aplicações em inferência estatística. Podemos obter o cálculo das probabilidades e da função densidade de probabilidade acumulada para diferentes valores de t da variável aleatória T.
